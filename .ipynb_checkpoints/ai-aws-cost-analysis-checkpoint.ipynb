```python
file_path = r"s3://aws-ai-cost-analysis/raw/Enterprise_AWS_Cost_Usage_Report-1.csv"
```

```python
df = pd.read_csv(file_path)
```

```python
df.head()
df.info
df.columns
```

```python
df['UsageDate'] = pd.to_datetime(df['UsageDate'], dayfirst=True)
df['UnblendedCost'] = pd.to_numeric(df['UnblendedCost'], errors='coerce')

df = df.dropna(subset=['UnblendedCost'])

df.info()
```

```python
service_summary = (
    df.groupby("Service")["UnblendedCost"]
    .sum()
    .reset_index()
    .sort_values(by="UnblendedCost", ascending=False)
)

service_summary
```

```python
df.head()
```

```python
import pandas as pd

# Convert date
df["UsageDate"] = pd.to_datetime(df["UsageDate"], dayfirst=True, errors="coerce")

# Convert numeric columns
df["UsageQuantity"] = pd.to_numeric(df["UsageQuantity"], errors="coerce")
df["UnblendedCost"] = pd.to_numeric(df["UnblendedCost"], errors="coerce")

# Remove invalid records
df = df.dropna(subset=["UsageDate", "UnblendedCost"])
df = df[df["UnblendedCost"] >= 0]

# Remove duplicates
df = df.drop_duplicates()

df.head()
```

```python
df["Year"] = df["UsageDate"].dt.year
df["Month"] = df["UsageDate"].dt.month
df["Day"] = df["UsageDate"].dt.day
df["MonthName"] = df["UsageDate"].dt.strftime("%B")
```

```python
daily_cost = (
    df.groupby("UsageDate")["UnblendedCost"]
    .sum()
    .reset_index()
    .rename(columns={"UnblendedCost": "TotalDailyCost"})
)
```

```python
df.head()
```

```python
service_cost = (
    df.groupby("Service")["UnblendedCost"]
    .sum()
    .reset_index()
    .sort_values(by="UnblendedCost", ascending=False)
)
```

```python
region_cost = (
    df.groupby("Region")["UnblendedCost"]
    .sum()
    .reset_index()
    .sort_values(by="UnblendedCost", ascending=False)
)
```

```python
account_cost = (
    df.groupby("LinkedAccount")["UnblendedCost"]
    .sum()
    .reset_index()
    .sort_values(by="UnblendedCost", ascending=False)
)
```

```python
usage_type_cost = (
    df.groupby("UsageType")["UnblendedCost"]
    .sum()
    .reset_index()
    .sort_values(by="UnblendedCost", ascending=False)
)
```

```python
mean_cost = daily_cost["TotalDailyCost"].mean()
std_cost = daily_cost["TotalDailyCost"].std()

daily_cost["ZScore"] = (
    (daily_cost["TotalDailyCost"] - mean_cost) / std_cost
)

anomalies = daily_cost[daily_cost["ZScore"].abs() > 2]
```

```python
summary = {
    "Total Cost": df["UnblendedCost"].sum(),
    "Top Service": service_cost.iloc[0]["Service"],
    "Top Region": region_cost.iloc[0]["Region"],
    "Top Account": account_cost.iloc[0]["LinkedAccount"],
    "Anomaly Days": len(anomalies)
}

summary

```

```python
import boto3
import pandas as pd
import io
from sqlalchemy import create_engine, text
from dotenv import load_dotenv
import os

# Load .env variables
load_dotenv()

# -----------------------------
# Step 1: S3 & DB Setup
# -----------------------------
bucket_name = os.getenv("S3_BUCKET") 
file_key = os.getenv("S3_PROCESSED_KEY") 

s3 = boto3.client('s3')
obj = s3.get_object(Bucket=bucket_name, Key=file_key)

cleaned_df = pd.read_csv(io.BytesIO(obj['Body'].read()))
print(cleaned_df.head())
```

```python
# -----------------------------
# Step 4: Push to RDS
# -----------------------------
table_name = "billing_data"

cleaned_df.to_sql(
    table_name,
    engine,
    if_exists='replace',  # overwrite table if exists
    index=False
)

print(f"Data pushed to RDS table '{table_name}' successfully ✅")

```

```python
# -----------------------------
# Step 5: Verify Table
# -----------------------------
with engine.connect() as connection:
    result = connection.execute(text(f"""
        SELECT table_name 
        FROM information_schema.tables 
        WHERE table_schema='public';
    """))
    tables = [row[0] for row in result]
    print("Tables in DB:", tables)
    
    # Optional: see first 5 rows
    sample = connection.execute(text(f"SELECT * FROM {table_name} LIMIT 5;"))
    for row in sample:
        print(row)
```

```python
# -----------------------------
# SQLAlchemy Dashboard Query Template
# -----------------------------
from sqlalchemy import create_engine, text
import pandas as pd
from dotenv import load_dotenv
import os

# Load .env variables
load_dotenv()

# Create engine
engine = create_engine(
    f"postgresql+psycopg2://{os.getenv('DB_USER')}:{os.getenv('DB_PASSWORD')}"
    f"@{os.getenv('DB_HOST')}:{os.getenv('DB_PORT')}/{os.getenv('DB_NAME')}"
)

print("Connected")
```

```python
from sqlalchemy import text
import pandas as pd

# Query total number of rows
query = text("SELECT COUNT(*) AS total_rows FROM billing_data;")

with engine.connect() as connection:
    result = connection.execute(query)
    for row in result:
        print("Total rows in table:", row[0])
```

```python
columns = pd.read_sql(
    "SELECT column_name FROM information_schema.columns WHERE table_name='billing_data';",
    engine
)
print(columns)
```

```python
from sqlalchemy import text

query = text('SELECT COUNT(*) AS total_rows FROM billing_data;')
with engine.connect() as connection:
    result = connection.execute(query)
    for row in result:
        print("Total rows:", row[0])
```

```python
# Aggregated
query =  """
    SELECT "Service", SUM("UnblendedCost") AS total_cost
    FROM billing_data
    GROUP BY "Service"
    ORDER BY total_cost DESC
    LIMIT 10;
    """
df_service_cost = pd.read_sql(query, engine)

# Preview results
print(df_service_cost)
```

```python
# Count total rows

from sqlalchemy import text
import pandas as pd

query = text('SELECT COUNT(*) AS total_rows FROM billing_data;')

with engine.connect() as connection:
    result = connection.execute(query)
    for row in result:
        print("Total rows in table:", row[0])
```

```python
# Total unblended cost
query = text('SELECT SUM("UnblendedCost") AS total_cost FROM billing_data;')

with engine.connect() as connection:
    result = connection.execute(query)
    for row in result:
        print("Total UnblendedCost:", row[0])
```

```python
# Top 5 services
query = text('''
SELECT "Service", SUM("UnblendedCost") AS service_cost
FROM billing_data
GROUP BY "Service"
ORDER BY service_cost DESC
LIMIT 5;
''')

with engine.connect() as connection:
    result = connection.execute(query)
    print("Top 5 services by total cost:")
    for row in result:
        print(row)
```

```python
# Daily cost trend (first 30 days)
query = text('''
SELECT "UsageDate", SUM("UnblendedCost") AS daily_cost
FROM billing_data
GROUP BY "UsageDate"
ORDER BY "UsageDate"
LIMIT 30;
''')

daily_cost_df = pd.read_sql(query, engine)
print(daily_cost_df)
```
